# Deep Dive: Exploring the TRAPPIST-1 System

TRAPPIST-1 is an ultra-cool red dwarf star 40 light-years away with 7 Earth-sized planets — the largest known system of terrestrial worlds. This notebook uses AstroAgent to investigate its habitability potential.

**What we'll do:**
1. Query the full system from NASA's Exoplanet Archive
2. Visualize all orbits in 3D with habitable zone overlay
3. Calculate precise habitable zone boundaries
4. Analyze which planets could support liquid water
5. Examine transit light curves for the most promising candidate

```python
from astroagent import AstroAgent

agent = AstroAgent(verbose=True)
```

## Step 1: Query the system

```python
result = agent.ask(
    "Query the NASA Exoplanet Archive for all confirmed planets in the TRAPPIST-1 system. "
    "Show me their orbital parameters, masses, radii, and equilibrium temperatures. "
    "Then calculate the habitable zone and visualize the full system in 3D."
)
```

## Step 2: Analyze habitability

```python
result = agent.ask(
    "Based on the data you just retrieved, which TRAPPIST-1 planets fall within "
    "the habitable zone? For each one, analyze whether it could realistically "
    "support liquid water, considering factors like tidal locking, stellar activity, "
    "and atmospheric retention."
)
print(result.text)
```

## Step 3: Examine the best candidate

```python
result = agent.ask(
    "Fetch the light curve data for TRAPPIST-1 and look for transit signals "
    "from planet e specifically. What does the transit depth tell us about "
    "the planet's size?"
)
```

## Step 4: Verify with evaluation harness

```python
from astroagent.eval import AccuracyEvaluator

evaluator = AccuracyEvaluator(tolerance_percent=5.0)
archive_data = result.data.get("query_exoplanet_archive", {})
planets = archive_data.get("planets", [])

if planets:
    evaluation = evaluator.evaluate_system("TRAPPIST-1", planets)
    print(evaluation.summary())
    print(f"\nDetailed results:")
    for r in evaluation.results:
        status = "PASS" if r.passed else "FAIL"
        print(f"  [{status}] {r.parameter}: expected={r.expected}, got={r.actual} (dev={r.deviation_percent}%)")
else:
    print("No planet data available for evaluation. Run Step 1 first.")
```
