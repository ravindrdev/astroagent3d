# Getting Started with AstroAgent 3D

This notebook demonstrates the core workflow: ask a question in natural language, get back real astronomical data with interactive 3D visualizations.

## Prerequisites

```bash
pip install -e .
export ANTHROPIC_API_KEY="your-key-here"
```

````python
from astroagent import AstroAgent

agent = AstroAgent(verbose=True)
````

## Your first query

Let's query a well-known exoplanetary system. The agent will:
1. Parse your natural language request
2. Decide which NASA APIs to call
3. Retrieve real data
4. Generate an interactive 3D visualization

````python
result = agent.ask("Show me all the planets in the TRAPPIST-1 system")
````

## Inspect the data

The agent's response includes both the text answer and the raw data it retrieved.

````python
import pandas as pd

planets = result.data.get("query_exoplanet_archive", {}).get("planets", [])
df = pd.DataFrame(planets)
df[["name", "orbital_period_days", "semi_major_axis_au", "mass_earth", "radius_earth", "equilibrium_temp_k"]]
````

## Follow-up questions

The agent maintains conversation context, so you can ask follow-up questions.

````python
result = agent.ask("Which of these planets is most likely to be habitable and why?")
print(result.text)
````
